# Age Outliers

```python
# FILTER INVALID AGES (0, 114, and unrealistic values)

# Check current min and max
print("Age range before:", user_df['Age'].min(), "to", user_df['Age'].max())

# Count invalid ages
invalid = user_df[(user_df['Age'] <= 5) | (user_df['Age'] >= 100)]
print(f"Records with impossible ages (<=5 or >=100): {len(invalid)}")

# Keep only realistic ages (6 to 99)
user_df = user_df[(user_df['Age'] > 5) & (user_df['Age'] < 100)]

# Confirm new range
print("Age range after:", user_df['Age'].min(), "to", user_df['Age'].max())
print(f"Total users after filtering: {len(user_df)}")
```

We cleaned up the age data by throwing out **928 fake or impossible ages** like people listed over 100 years old. Instead of ages going all the way from **0 to 114**, the real ages now fall between **9 and 97**. After tossing out those bad records, we are left with **4,447 clean user profiles** that we can actually trust

# Merge
**Merge:** Combines the `Viewership` and `User Profiles` datasets into a single unified dataframe containing all columns from both sources.

Merging these datasets pairs user demographics (such as age, gender, and province) with viewing behavior (such as watch duration and channels watched). This enables comprehensive analysis to identify who is watching what and uncover the key factors influencing consumption patterns.

```python
# FIX: Drop the redundant/conflicting 'userid' column from viewership before merging
if 'userid' in view_df.columns:
    view_df = view_df.drop(columns=['userid'])

# Ensure Duration_seconds exists
if 'Duration_seconds' not in view_df.columns:
    view_df['Duration_str'] = view_df['Duration 2'].astype(str)
    view_df['Duration_str'] = view_df['Duration_str'].replace({'nan': '00:00:00', 'None': '00:00:00'})
    view_df['Duration_seconds'] = pd.to_timedelta(view_df['Duration_str']).dt.total_seconds()

# Merge everything on UserID (inner join keeps only matching records)
merged_df = view_df.merge(user_df, on='UserID', how='inner')

print("Merged dataframe shape:", merged_df.shape)
```

* Age Group vs Average Watch Duration

Determine which certain age groups watch longer.

it will also help target age groups with personalised recommendations.

```python
# Merge viewership with user profiles to get age group info
merged_df = view_df.merge(user_df[['UserID', 'Age_Group', 'Gender', 'Province']], left_on='UserID', right_on='UserID', how='inner')

# Filter out rows with missing age group or zero duration
merged_df = merged_df[merged_df['Age_Group'].notna()]
merged_df = merged_df[merged_df['Duration_seconds'] > 0]

# Group by age group and calculate average duration in minutes
age_duration = merged_df.groupby('Age_Group')['Duration_seconds'].mean().reset_index()
age_duration['Avg_Duration_Minutes'] = (age_duration['Duration_seconds'] / 60).round(1)

# Ensure correct order
age_order = ['0-17', '18-24', '25-34', '35-44', '45-54', '55+']
age_duration['Age_Group'] = pd.Categorical(age_duration['Age_Group'], categories=age_order, ordered=True)
age_duration = age_duration.sort_values('Age_Group')

# Bar chart
fig = px.bar(age_duration,
             x='Age_Group',
             y='Avg_Duration_Minutes',
             title='Average Watch Duration by Age Group',
             labels={'Age_Group': 'Age Group', 'Avg_Duration_Minutes': 'Average Duration (minutes)'},
             color='Age_Group',
             color_discrete_sequence=px.colors.qualitative.Set2)

fig.show()
```

The chart shows that users aged 45–54 have the highest average watch duration, followed by the 25–34 and 35–44 age groups. In contrast, the 55+ age group has the shortest average watch duration. This suggests that middle-aged users are the most engaged, making them an important audience for targeted content and marketing.

meaning

Decline meaning fewer viewing sessions or shorter watch time compared to the previous month.

This will help us to identify seasonal trends and plan content releases or marketing campaigns during high and low consumption months.

```python
import plotly.express as px
import pandas as pd

# Ensure SA_RecordDate exists (SA time column)
if 'SA_RecordDate' not in view_df.columns:
    view_df['SA_RecordDate'] = pd.to_datetime(view_df['RecordDate2']) + pd.Timedelta(hours=2)

# Use merged_df if you have it; otherwise work with view_df
df = merged_df.copy() if 'merged_df' in globals() else view_df.copy()

# Remove zero/negative durations (if using total duration)
df = df[df['Duration_seconds'] > 0]

# Create a Month-Year column for grouping
df['Month_Year'] = df['SA_RecordDate'].dt.to_period('M').astype(str)

# Option 1: Total sessions per month
monthly_sessions = df.groupby('Month_Year').size().reset_index(name='Sessions')
monthly_sessions = monthly_sessions.sort_values('Month_Year')

# Option 2: Total watch duration (hours) per month
monthly_duration = df.groupby('Month_Year')['Duration_seconds'].sum().reset_index()
monthly_duration['Total_Hours'] = (monthly_duration['Duration_seconds'] / 3600).round(1)
monthly_duration = monthly_duration.sort_values('Month_Year')

# Choose which metric to plot: sessions or hours
# Option A: Sessions per month
fig1 = px.line(monthly_sessions,
               x='Month_Year',
               y='Sessions',
               title='Monthly Viewing Sessions (Consumption Trend)',
               labels={'Month_Year': 'Month', 'Sessions': 'Number of Sessions'},
               markers=True)

fig1.show()

# Option B: Total watch hours per month
fig2 = px.line(monthly_duration,
               x='Month_Year',
               y='Total_Hours',
               title='Monthly Total Watch Time (Consumption Trend)',
               labels={'Month_Year': 'Month', 'Total_Hours': 'Total Hours Watched'},
               markers=True)

fig2.show()
```

Both graphs show
Viewing sessions and total watch time increased from January to March 2016, with March having the highest consumption. Both dropped to zero in April because no viewing activity is recorded for April in the dataset.

```python
import plotly.express as px
import pandas as pd

# Ensure SA_RecordDate exists (SA time)
if 'SA_RecordDate' not in view_df.columns:
    view_df['SA_RecordDate'] = pd.to_datetime(view_df['RecordDate2']) + pd.Timedelta(hours=2)

# Add Day_Name if not already present
if 'Day_Name' not in view_df.columns:
    view_df['Day_Name'] = view_df['SA_RecordDate'].dt.day_name()

# Use view_df or merged_df (if you have it)
df = merged_df.copy() if 'merged_df' in globals() else view_df.copy()

# Count sessions per day of the week
daily_sessions = df.groupby('Day_Name').size().reset_index(name='Sessions')

# Ensure correct weekday order
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_sessions['Day_Name'] = pd.Categorical(daily_sessions['Day_Name'], categories=day_order, ordered=True)
daily_sessions = daily_sessions.sort_values('Day_Name')

# Find the lowest day
lowest_day = daily_sessions.loc[daily_sessions['Sessions'].idxmin()]
print(f"\n📉 Lowest Viewing Day: {lowest_day['Day_Name']} with {lowest_day['Sessions']} sessions")

# Bar chart showing all days
fig = px.bar(daily_sessions,
             x='Day_Name',
             y='Sessions',
             title='Sessions per Day of Week',
             labels={'Day_Name': 'Day', 'Sessions': 'Number of Sessions'},
             color='Day_Name',
             color_discrete_sequence=px.colors.qualitative.Set2)

fig.add_annotation(
    x=lowest_day['Day_Name'],
    y=lowest_day['Sessions'],
    text=f"Lowest: {lowest_day['Sessions']}",
    showarrow=True,
    arrowhead=1,
    ax=0,
    ay=-30
)

fig.show()
```

# Compare Weekend vs Weekday
Weekend behaviour differs from weekdays.

It will help recommend different content for weekends.

```python
import plotly.express as px
import pandas as pd

# Ensure SA_RecordDate and Duration_seconds exist
if 'SA_RecordDate' not in view_df.columns:
    view_df['SA_RecordDate'] = pd.to_datetime(view_df['RecordDate2']) + pd.Timedelta(hours=2)

if 'Duration_seconds' not in view_df.columns:
    view_df['Duration_str'] = view_df['Duration 2'].astype(str)
    view_df['Duration_str'] = view_df['Duration_str'].replace({'nan': '00:00:00', 'None': '00:00:00'})
    view_df['Duration_seconds'] = pd.to_timedelta(view_df['Duration_str']).dt.total_seconds()

# Add Day_Name if not exists
if 'Day_Name' not in view_df.columns:
    view_df['Day_Name'] = view_df['SA_RecordDate'].dt.day_name()

# Use merged_df if available, else view_df
df = merged_df.copy() if 'merged_df' in globals() else view_df.copy()

# Add Weekend flag to the 'df' DataFrame that will be used for calculations
df['Is_Weekend'] = df['Day_Name'].isin(['Saturday', 'Sunday'])

# Remove invalid durations
df = df[df['Duration_seconds'] > 0]

# ----- Calculate Metrics -----

# 1. Sessions per day type
weekend_sessions = df[df['Is_Weekend'] == True].shape[0]
weekday_sessions = df[df['Is_Weekend'] == False].shape[0]

# 2. Average duration per day type
weekend_avg_duration = df[df['Is_Weekend'] == True]['Duration_seconds'].mean() / 60
weekday_avg_duration = df[df['Is_Weekend'] == False]['Duration_seconds'].mean() / 60

# Create summary dataframe for plotting
summary_df = pd.DataFrame({
    'Day_Type': ['Weekday', 'Weekend'],
    'Sessions': [weekday_sessions, weekend_sessions],
    'Avg_Duration_Minutes': [weekday_avg_duration.round(1), weekend_avg_duration.round(1)]
})

print("="*50)
print("WEEKEND vs WEEKDAY COMPARISON")
print("="*50)
print(f"📊 Weekday Sessions: {weekday_sessions:,}")
print(f"📊 Weekend Sessions: {weekend_sessions:,}")
print(f"\n⏱️ Weekday Average Duration: {weekday_avg_duration:.1f} minutes")
print(f"⏱️ Weekend Average Duration: {weekend_avg_duration:.1f} minutes")

# ----- Bar Chart: Sessions -----
fig1 = px.bar(summary_df,
              x='Day_Type',
              y='Sessions',
              title='Sessions: Weekend vs Weekday',
              labels={'Day_Type': '', 'Sessions': 'Number of Sessions'},
              color='Day_Type',
              color_discrete_sequence=['#3366CC', '#FF6B6B'],
              text='Sessions')

fig1.update_traces(textposition='outside')
fig1.show()

# ----- Bar Chart: Average Duration -----
fig2 = px.bar(summary_df,
              x='Day_Type',
              y='Avg_Duration_Minutes',
              title='Average Duration: Weekend vs Weekday',
              labels={'Day_Type': '', 'Avg_Duration_Minutes': 'Average Duration (minutes)'},
              color='Day_Type',
              color_discrete_sequence=['#3366CC', '#FF6B6B'],
              text='Avg_Duration_Minutes')

fig2.update_traces(textposition='outside')
fig2.show()
```

Weekdays have more viewing sessions 6,197 than weekends 2,709, but weekends have a slightly higher average watch duration 10.6 minutes vs 9.8 minutes. This means users watch more often during weekdays, but stay slightly longer per session on weekends.

We could focus on more frequent content during weekdays and longer engaging content on weekends.
